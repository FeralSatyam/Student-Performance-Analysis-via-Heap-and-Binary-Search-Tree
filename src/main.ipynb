```python
#Loading CSV Data
with open ('../data/StudentPerformanceFactors.csv', 'r') as f:
    lines = f.readlines()
    for line in lines:
        row = line.strip().split(',')
    #Converting items in the list to integers
    numeric_index = [0, 1, 5, 6, 9, 14, 19]
    for i in numeric_index:
        row[i] = int(row[i])
```

```python
#Index of each row of dataset
# 0. Hours_Studied -> int
# 1. Attendance -> int
# 2. Parental_Involvement -> str -> low, medium, high
# 3. Access_to_Resources -> str -> low, medium, high
# 4. Extracurricular_Activities -> yes, no
# 5. Sleep_Hours -> int
# 6. Previous_Scores -> int
# 7. Motivation_Level -> str -> low, medium, high
# 8. Internet_Access -> yes, no
# 9. Tutoring_Sessions -> int
# 10. Family_Income -> str -> low, medium, high
# 11. Teacher_Quality -> str -> low, medium, high
# 12. School_Type -> str -> public, private
# 13. Peer_Influence -> str -> positive, negative, neutral
# 14. Physical_Activity -> int
# 15. Learning_Disabilities 
# 16. Parental_Education_Level
# 17. Distance_from_Home
# 18. Gender
# 19. Exam_Score
```

```python
#Creating a dictionary with all the students
with open ('../data/StudentPerformanceFactors.csv', 'r') as f:
    first_lines = f.readline()
    headers = first_lines.strip().split(',')
    all_students = []

    for line in f:
        values = line.strip().split(',')
        student = {}
        for i in range(len(headers)):
            header = headers[i]
            value = values[i]
             #Converting items in the list to integers
            numeric_column = ['Hours_Studied', 'Attendance', 'Sleep_Hours', 'Previous_Scores', 'Tutoring_Sessions', 'Physical_Activity', 'Exam_Score']
            if header in numeric_column:
                value = int(value)
            student[header] = value
        all_students.append(student)
```
