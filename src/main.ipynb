```python
#Searching for min and max score within the dataset
def search_min_max_score(all_students):
    min_score = 100
    max_score = 0
    for i in range(len(all_students)):
        if all_students[i]['Exam_Score'] < min_score:
            min_score = all_students[i]['Exam_Score']
        if all_students[i]['Exam_Score'] > max_score:
            max_score = all_students[i]['Exam_Score']
    return min_score, max_score
```

```python
#Identifying min and max exam score among the students
min_score, max_score = search_min_max_score(all_students)
print(f"Min score: {min_score}, Max score: {max_score}")
```

```python
#Calculating average attendence of all students
def calculate_average_attendence(all_students):
    total_attendence = 0
    no_of_students = len(all_students)
    for i in range(no_of_students):
        total_attendence += all_students[i]['Attendance']
    average_attendence = total_attendence / no_of_students
    return average_attendence
```

```python
#Average Attendence of students within the dataset
average_attendence = calculate_average_attendence(all_students)
print(f"Average Attendence of students: {average_attendence}")
```

```python
#Calculating number of students with different motivation levels
def calculate_motivation_level(all_students):
    no_of_students = len(all_students)
    low_motivation = 0
    medium_motivation = 0
    high_motivation = 0
    for i in range(no_of_students):
        if all_students[i]['Motivation_Level'] == 'Low':
            low_motivation += 1
        elif all_students[i]['Motivation_Level'] == 'Medium':
            medium_motivation += 1
        else:
            high_motivation += 1
    return low_motivation, medium_motivation, high_motivation
```

```python
low_motivation, medium_motivation, high_motivation = calculate_motivation_level(all_students)
print(f"No of students with \nLow motivation: {low_motivation}\nMedium Motivation: {medium_motivation}\nHigh Motivation: {high_motivation}")
```

```python
#Calculating risk of each student
def calculate_risk(all_students):
    all_students_risk = []
    no_of_students = len(all_students)
    for i in range(no_of_students):
        risk = 0
        if all_students[i]['Attendance'] >= 80:
            risk += 0
        else:
            risk += (100 - (all_students[i]['Attendance'])) * 0.4
        if all_students[i]['Hours_Studied'] >= 10:
            risk += 0
        else:
            risk += (10 - (all_students[i]['Hours_Studied'])) * 1.5
        if all_students[i]['Previous_Scores'] >= 90:
            risk += 0
        else:
            risk += (100 - (all_students[i]['Previous_Scores'])) * 0.3
        if all_students[i]['Tutoring_Sessions'] >= 3:
            risk += 0
        else:
            risk += (3 - (all_students[i]['Tutoring_Sessions'])) * 2.0
        if all_students[i]['Motivation_Level'] == 'high_motivation':
            risk += 0
        elif all_students[i]['Motivation_Level'] == 'medium_motivation':
            risk += 5
        else:
            risk += 10
        all_students_risk.append(risk)
    return all_students_risk
```

```python
all_students_risk = calculate_risk(all_students)
```

```python
#Creating a tuple with student risk, student index and student dictionary
student_info = []

no_of_students = len(all_students)
for i in range(no_of_students):
    curr_student_index = i
    curr_student_risk = all_students_risk[i]
    curr_student_dict = all_students[i]
    student_info.append((curr_student_risk, curr_student_index, curr_student_dict))
student_info = list(student_info)
```

```python
#Implementing heap shift down
def shift_down(heap, index, size):
    largest = index
    left = (index * 2) + 1
    right = (index * 2) +  2

    if left < size and heap[left][0] > heap[largest][0]:
        largest = left
    if right < size and heap[right][0] > heap[largest][0]:
        largest = right
    if largest != index:
        temp = heap[index]
        heap[index] = heap[largest]
        heap[largest] = temp
        shift_down(heap, largest, size)

def heapify(heap):
    size = len(heap)
    if size <= 1:
        return
    last_parent = (size // 2) - 1
    for i in range(last_parent, -1, -1):
        shift_down(heap, i, size)

def heap_pop(heap):
    if len(heap) == 0:
        return None
    max_item = heap[0]
    heap[0] = heap[-1]
    heap.pop()
    if len(heap) > 0:
        shift_down(heap, 0, len(heap))
    return max_item
```

```python
class Node:
    def __init__(self, exam_score, student_data):
        self.key = exam_score
        self.data = student_data
        self.left = None
        self.right = None

def bst_insert(root, node):
    if root is None:
        return node
    if node.key < root.key:
        root.left = bst_insert(root.left, node)
    else:
        root.right = bst_insert(root.right, node)
    return root

def range_query(root, low, high, result):
    if root is None:
        return
    if low < root.key:
        range_query(root.left, low, high, result)
    if low <= root.key <= high:
        result.append((root.key, root.data))
    if root.key < high:
        range_query(root.right, low, high, result)

def inorder_collect(root, result):
    if root is None:
        return
    inorder_collect(root.left, result)
    result.append(root.key)
    inorder_collect(root.right, result)

# Build BST
print("Building BST...")
root = None
for student in all_students:
    root = bst_insert(root, Node(student['Exam_Score'], student))
print("BST built successfully")
```

```python
def get_top_risk(k):
    """Get top k at-risk students using heap"""
    temp_heap = student_info[:]
    heapify(temp_heap)
    results = []
    for i in range(min(k, len(temp_heap))):
        results.append(heap_pop(temp_heap))
    return results

def show_top_risk_students(n):
    print(f"\n--- TOP {n} AT-RISK STUDENTS ---")
    print("(Go to CSV Row number to find the student in the original file)")
    print("-" * 85)
    top_k = get_top_risk(n)
    for rank, item in enumerate(top_k, 1):
        risk = item[0]
        index = item[1]           # Index in all_students (0 = CSV row 2)
        csv_row = index + 2        # CSV row number (because row 1 is headers)
        student = item[2]
        print(f"{rank:2}. CSV Row: {csv_row:4} | Risk: {risk:6.1f} | Attendance: {student['Attendance']:3}% | Study: {student['Hours_Studied']:2}hrs | Previous: {student['Previous_Scores']:3} | Motivation: {student['Motivation_Level']}")

# Example usage
show_top_risk_students(5)
```
