```python
#Creating a dictionary with all the students
with open ('../data/StudentPerformanceFactors.csv', 'r') as f:
    first_lines = f.readline()
    headers = first_lines.strip().split(',')
    all_students = []

    for line in f:
        values = line.strip().split(',')
        student = {}
        for i in range(len(headers)):
            header = headers[i]
            value = values[i]
             #Converting items in the list to integers
            numeric_column = ['Hours_Studied', 'Attendance', 'Sleep_Hours', 'Previous_Scores', 'Tutoring_Sessions', 'Physical_Activity', 'Exam_Score']
            if header in numeric_column:
                value = int(value)
            student[header] = value
        all_students.append(student)
```

# Performing EDA

```python
#Searching for min and max score within the dataset
def search_min_max_score(all_students):
    min_score = 100
    max_score = 0
    for i in range(len(all_students)):
        if all_students[i]['Exam_Score'] < min_score:
            min_score = all_students[i]['Exam_Score']
        if all_students[i]['Exam_Score'] > max_score:
            max_score = all_students[i]['Exam_Score']
    return min_score, max_score
```

```python
#Identifying min and max exam score among the students
min_score, max_score = search_min_max_score(all_students)
print(f"Min score: {min_score}, Max score: {max_score}")
```

```python
#Calculating average attendence of all students
def calculate_average_attendence(all_students):
    total_attendence = 0
    no_of_students = len(all_students)
    for i in range(no_of_students):
        total_attendence += all_students[i]['Attendance']
    average_attendence = total_attendence / no_of_students
    return average_attendence
```

```python
#Average Attendence of students within the dataset
average_attendence = calculate_average_attendence(all_students)
print(f"Average Attendence of students: {average_attendence}")
```

```python
#Calculating number of students with different motivation levels
def calculate_motivation_level(all_students):
    no_of_students = len(all_students)
    low_motivation = 0
    medium_motivation = 0
    high_motivation = 0
    for i in range(no_of_students):
        if all_students[i]['Motivation_Level'] == 'Low':
            low_motivation += 1
        elif all_students[i]['Motivation_Level'] == 'Medium':
            medium_motivation += 1
        else:
            high_motivation += 1
    return low_motivation, medium_motivation, high_motivation
```

```python
low_motivation, medium_motivation, high_motivation = calculate_motivation_level(all_students)
print(f"No of students with \nLow motivation: {low_motivation}\nMedium Motivation: {medium_motivation}\nHigh Motivation: {high_motivation}")
```
